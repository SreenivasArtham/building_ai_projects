```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
requests = [1000, 1500, 1300, 2100, 2500]

fig, ax = plt.subplots()

ax.plot(days, requests, marker="o")

ax.set_title("Daily API Requests")
ax.set_xlabel("Day")
ax.set_ylabel("Requests")

plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
requests = [1000, 1500, 1300, 2100, 2500]
errors = [12, 25, 18, 40, 32]

# 1 Figure with 1 row and 2 columns (2 separate Axes)
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))

# Overall title for the entire Figure
fig.suptitle("Weekly API Traffic & Errors", fontsize=14, fontweight="bold")

# --- First Axes (Left) ---
ax1.plot(days, requests, marker="v", color="red")
ax1.set_title("Daily API Requests")
ax1.set_xlabel("Day")
ax1.set_ylabel("Requests")

# --- Second Axes (Right) ---
ax2.bar(days, errors, color="blue")
ax2.set_title("Daily API Errors")
ax2.set_xlabel("Day")
ax2.set_ylabel("Error Count")

# Prevents axis labels from overlapping between the two plots
plt.tight_layout()
plt.show()
```

```python
latency_ms = [20, 30, 25, 40, 50, 55, 60, 65,
               70, 75, 80, 90, 100, 120, 150]

fig, ax = plt.subplots()

ax.hist(latency_ms, bins=10)

ax.set_title("API Latency Distribution")
ax.set_xlabel("Latency (ms)")
ax.set_ylabel("Frequency")

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot([1, 2, 3], [10, 20, 30])
axes[0, 0].set_title("Line Chart")

axes[0, 1].bar(["A", "B", "C"], [10, 20, 30])
axes[0, 1].set_title("Bar Chart")

axes[1, 0].hist([10, 20, 20, 30, 40], bins=4)
axes[1, 0].set_title("Histogram")

axes[1, 1].scatter([1, 2, 3], [30, 20, 50])
axes[1, 1].set_title("Scatter")

plt.tight_layout()
plt.show()
```

```python
#Matplotlib with Pandas
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({ "month": ["Jan", "Feb", "Mar", "Apr"],
                    "sales": [10000, 15000, 12000, 20000]
                })
df.plot(x="month",y="sales",kind="line",marker="o")
plt.title("Monthly Sales")
plt.ylabel("Sales")
plt.show()
```

```python
df = pd.DataFrame({
    "department": ["IT", "IT", "HR", "HR", "Finance"],
    "salary": [90000, 110000, 60000, 70000, 80000]
})

avg_salary = df.groupby("department")["salary"].mean()

avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.ylabel("Average Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Plot data series with labels
plt.plot(x, y1, label="Sine Wave", color="blue", linewidth=2)
plt.plot(x, y2, label="Cosine Wave", color="red", linestyle="--", linewidth=2)

# Add legend and styled gridlines
plt.legend(loc="upper right", title="Functions")
plt.grid(True, linestyle="--", alpha=0.5, color="gray")

plt.title("Sine and Cosine Waves")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```
